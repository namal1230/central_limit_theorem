```python
import numpy as np
import seaborn as sns
import statistics as stat
```

```python
x = np.random.normal(size=1000)
```

```python
sns.displot(x)
```

```python
sns.displot(x,kde=False)
```

```python
x_sample = np.random.choice(x, size=10, replace=False)
x_sample
```

```python
stat.mean(x_sample)
```

```python
def sample_mean_calculater(population_array,sample_size,n_samples):
    sample_means = []
    for i in range (n_samples):
        sample = np.random.choice(population_array, size=sample_size, replace=False)
        sample_mean = stat.mean(sample)
        sample_means.append(sample_mean)
    return sample_means
```

```python
_ = sns.displot(sample_mean_calculater(x,10,10))
```

```python
_ = sns.displot(sample_mean_calculater(x,10,1000))
```

```python
_ = sns.displot(sample_mean_calculater(x,1000,1000))
```

```python
from scipy.stats import skewnorm
```

```python
s = skewnorm.rvs(12, size=10000)
```

```python
_ = sns.distplot(sample_mean_calculater(s,10,1000))
```
